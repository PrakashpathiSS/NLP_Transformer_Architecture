# Transformer Text Generation — Mini Project

This notebook converts the previous Transformer translation structure into a **decoder-only Transformer text generation model**.

### Main concept

**Previous tokens → Transformer → next-token prediction**

Example:

`Artificial intelligence` → `helps`

Then:

`Artificial intelligence helps` → `analyze`

The model generates text one token at a time.


## 1. Imports

```python
# ============================================================
# 1. IMPORTS
# ============================================================

import numpy as np
import tensorflow as tf
import pandas as pd

from tensorflow.keras.layers import (
    Embedding,
    Dense,
    Dropout,
    LayerNormalization,
    MultiHeadAttention,
)
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

print("TensorFlow:", tf.__version__)
```

## 2. Load the Text Generation Dataset

```python
# ============================================================
# 2. LOAD DATASET
# ============================================================

DATASET_PATH = "transformer_text_generation_500_rows(2).csv"

dataset = pd.read_csv(DATASET_PATH)

texts = dataset["text"].astype(str).tolist()

print("Training texts:", len(texts))

for i in range(min(5, len(texts))):
    print(f"{i + 1}. {texts[i]}")
    print()
```

## 3. Tokenizer

For text generation, we use **one tokenizer** because the model is learning one language/text stream.

There is no English tokenizer + Tamil tokenizer here.

The tokenizer converts:

`Artificial intelligence is useful`

into:

`[token_id, token_id, token_id, token_id]`


```python
# ============================================================
# 3. TOKENIZER
# ============================================================

MAX_VOCAB_SIZE = 5000

tokenizer = Tokenizer(
    num_words=MAX_VOCAB_SIZE,
    filters="",
    oov_token="<unk>",
)

tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)

vocab_size = min(
    len(tokenizer.word_index) + 1,
    MAX_VOCAB_SIZE,
)

print("Actual vocabulary:", len(tokenizer.word_index) + 1)
print("Vocabulary used by model:", vocab_size)
```

## 4. Add `<eos>` and Create Token Sequences

```python
# ============================================================
# 4. ADD EOS AND CREATE SEQUENCES
# ============================================================

EOS_TOKEN = "<eos>"

# Add <eos> so the model can learn when a generated sentence should end.
texts_with_eos = [
    text + " " + EOS_TOKEN
    for text in texts
]

tokenizer.fit_on_texts([EOS_TOKEN])

sequences = tokenizer.texts_to_sequences(texts_with_eos)

# Keep only tokens inside the model vocabulary.
sequences = [
    [token if token < vocab_size else tokenizer.word_index["<unk>"]
     for token in seq]
    for seq in sequences
]

print("Example text:")
print(texts_with_eos[0])
print()
print("Token IDs:")
print(sequences[0])
```

## 5. Create Input → Next Token Target

```python
# ============================================================
# 5. INPUT / TARGET CREATION
# ============================================================

# Example:
#
# Tokens:
# [Artificial, intelligence, is, useful, <eos>]
#
# Input:
# [Artificial, intelligence, is, useful]
#
# Target:
# [intelligence, is, useful, <eos>]
#
# The model learns to predict the NEXT token.

CONTEXT_LENGTH = 32

input_sequences = []
target_sequences = []

for seq in sequences:

    # Ignore sequences that are too short.
    if len(seq) < 2:
        continue

    # Create multiple training windows from longer text.
    for start in range(0, len(seq) - 1, CONTEXT_LENGTH):

        chunk = seq[start:start + CONTEXT_LENGTH + 1]

        if len(chunk) < 2:
            continue

        input_part = chunk[:-1]
        target_part = chunk[1:]

        input_sequences.append(input_part)
        target_sequences.append(target_part)

# Pad every training sequence to the same context length.
input_sequences = pad_sequences(
    input_sequences,
    maxlen=CONTEXT_LENGTH,
    padding="post",
    truncating="pre",
)

target_sequences = pad_sequences(
    target_sequences,
    maxlen=CONTEXT_LENGTH,
    padding="post",
    truncating="pre",
)

# Ignore padded target positions when calculating loss/accuracy.
sample_weight = (
    target_sequences != 0
).astype("float32")

print("Input shape:", input_sequences.shape)
print("Target shape:", target_sequences.shape)
```

## 6. Positional Embedding

```python
# ============================================================
# 6. POSITIONAL EMBEDDING
# ============================================================

class PositionalEmbedding(tf.keras.layers.Layer):

    def __init__(
        self,
        vocab_size,
        sequence_length,
        d_model,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.token_embedding = Embedding(
            vocab_size,
            d_model,
            mask_zero=True,
        )

        self.position_embedding = Embedding(
            sequence_length,
            d_model,
        )

    def call(self, inputs):

        length = tf.shape(inputs)[-1]

        positions = tf.range(
            start=0,
            limit=length,
            delta=1,
        )

        positions = self.position_embedding(positions)
        tokens = self.token_embedding(inputs)

        return tokens + positions
```

## 7. Decoder-Only Transformer Block

Unlike the previous translation model, there is **no encoder and no cross-attention**.

The block only uses:

1. Causal self-attention
2. Residual connection + LayerNorm
3. Feed-forward network
4. Residual connection + LayerNorm

Causal attention means a token can look at the **past**, but not the future.


```python
# ============================================================
# 7. DECODER-ONLY TRANSFORMER BLOCK
# ============================================================

class TransformerDecoderBlock(tf.keras.layers.Layer):

    def __init__(
        self,
        d_model,
        num_heads,
        ff_dim,
        dropout=0.1,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.self_attention = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads,
            dropout=dropout,
        )

        self.ffn = tf.keras.Sequential([
            Dense(ff_dim, activation="relu"),
            Dense(d_model),
        ])

        self.norm1 = LayerNormalization(epsilon=1e-6)
        self.norm2 = LayerNormalization(epsilon=1e-6)

        self.dropout1 = Dropout(dropout)
        self.dropout2 = Dropout(dropout)

    def call(
        self,
        x,
        training=False,
        attention_mask=None,
    ):

        # Causal self-attention.
        attn_output = self.self_attention(
            query=x,
            key=x,
            value=x,
            attention_mask=attention_mask,
            training=training,
        )

        x = self.norm1(
            x + self.dropout1(
                attn_output,
                training=training,
            )
        )

        # Feed-forward network.
        ffn_output = self.ffn(
            x,
            training=training,
        )

        x = self.norm2(
            x + self.dropout2(
                ffn_output,
                training=training,
            )
        )

        return x
```

## 8. Causal Mask

```python
# ============================================================
# 8. CAUSAL MASK
# ============================================================

def causal_mask(length):

    # 1 = allowed to attend
    # 0 = blocked

    i = tf.range(length)[:, None]
    j = tf.range(length)[None, :]

    return tf.cast(
        i >= j,
        tf.int32,
    )

# Example for 4 tokens:
#
# [[1, 0, 0, 0],
#  [1, 1, 0, 0],
#  [1, 1, 1, 0],
#  [1, 1, 1, 1]]
#
# Token 1 cannot see tokens 2, 3, 4.
# Token 2 cannot see tokens 3, 4.
```

## 9. Decoder-Only Transformer

```python
# ============================================================
# 9. DECODER-ONLY TRANSFORMER
# ============================================================

class Transformer(tf.keras.Model):

    def __init__(
        self,
        vocab_size,
        max_sequence_length,
        d_model=64,
        num_heads=4,
        ff_dim=128,
        dropout=0.1,
        **kwargs,
    ):
        super().__init__(**kwargs)

        self.embedding = PositionalEmbedding(
            vocab_size,
            max_sequence_length,
            d_model,
        )

        self.decoder = TransformerDecoderBlock(
            d_model,
            num_heads,
            ff_dim,
            dropout,
        )

        # One probability/logit for every vocabulary token.
        self.output_layer = Dense(vocab_size)

    def call(self, inputs, training=False):

        token_ids = inputs

        x = self.embedding(token_ids)

        # Padding mask.
        padding_mask = tf.cast(
            token_ids != 0,
            tf.int32,
        )[:, None, :]

        # Causal mask.
        sequence_length = tf.shape(token_ids)[1]

        cmask = causal_mask(
            sequence_length
        )[None, :, :]

        # Combine padding + causal masks.
        attention_mask = (
            cmask *
            tf.cast(
                padding_mask[:, None, :],
                tf.int32,
            )
        )

        x = self.decoder(
            x,
            training=training,
            attention_mask=attention_mask,
        )

        return self.output_layer(x)
```

## 10. Build the Small Text Generation Transformer

```python
# ============================================================
# 10. BUILD MODEL
# ============================================================

D_MODEL = 64
NUM_HEADS = 4
FF_DIM = 128
DROPOUT = 0.1

transformer = Transformer(
    vocab_size=vocab_size,
    max_sequence_length=CONTEXT_LENGTH,
    d_model=D_MODEL,
    num_heads=NUM_HEADS,
    ff_dim=FF_DIM,
    dropout=DROPOUT,
)

# Build the model once.
dummy_input = tf.constant(
    input_sequences[:1],
    dtype=tf.int32,
)

dummy_logits = transformer(
    dummy_input,
    training=False,
)

transformer.summary()

print("Logits shape:", dummy_logits.shape)
```

## 11. Compile

The target is an integer token ID.

Therefore we use:

`SparseCategoricalCrossentropy(from_logits=True)`

At every position the model predicts one probability/logit for every vocabulary token.


```python
# ============================================================
# 11. COMPILE
# ============================================================

loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=True
)

transformer.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss=loss_fn,
    weighted_metrics=["accuracy"],
)
```

## 12. Train

```python
# ============================================================
# 12. TRAIN
# ============================================================

EPOCHS = 30
BATCH_SIZE = 16

history = transformer.fit(
    input_sequences,
    target_sequences,
    sample_weight=sample_weight,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    shuffle=True,
    verbose=1,
)
```

## 13. Autoregressive Text Generation

During inference there is no target sentence.

The model starts with the user's prompt and repeatedly predicts the next token.

Example:

`Artificial intelligence`

→ predict `helps`

→ `Artificial intelligence helps`

→ predict `analyze`

→ `Artificial intelligence helps analyze`

This is called **autoregressive generation**.


```python
# ============================================================
# 13. TEXT GENERATION
# ============================================================

eos_id = tokenizer.word_index[EOS_TOKEN]

index_to_word = {
    index: word
    for word, index in tokenizer.word_index.items()
    if index < vocab_size
}

def generate_text(
    prompt,
    max_new_tokens=30,
):

    # Convert prompt into token IDs.
    prompt_ids = tokenizer.texts_to_sequences(
        [prompt]
    )[0]

    if not prompt_ids:
        return prompt

    # Replace IDs outside the model vocabulary with <unk>.
    unk_id = tokenizer.word_index["<unk>"]

    prompt_ids = [
        token if token < vocab_size else unk_id
        for token in prompt_ids
    ]

    generated_ids = prompt_ids.copy()

    for _ in range(max_new_tokens):

        # Use only the latest context window.
        context_ids = generated_ids[-CONTEXT_LENGTH:]

        model_input = pad_sequences(
            [context_ids],
            maxlen=CONTEXT_LENGTH,
            padding="post",
            truncating="pre",
        )

        logits = transformer(
            tf.constant(
                model_input,
                dtype=tf.int32,
            ),
            training=False,
        )

        # Number of actual tokens in this context.
        current_length = len(context_ids)

        # Predict the token after the last real token.
        next_logits = logits[
            0,
            current_length - 1,
        ]

        next_id = int(
            tf.argmax(next_logits).numpy()
        )

        # Stop when EOS is generated.
        if next_id == eos_id or next_id == 0:
            break

        generated_ids.append(next_id)

    words = []

    for token_id in generated_ids:

        if token_id == eos_id or token_id == 0:
            break

        words.append(
            index_to_word.get(
                token_id,
                "<unk>"
            )
        )

    return " ".join(words)
```

## 14. Test Text Generation

```python
# ============================================================
# 14. TEST GENERATION
# ============================================================

test_prompts = [
    "Artificial intelligence",
    "Machine learning",
    "Transformers",
    "Natural language processing",
    "Python programming",
]

for prompt in test_prompts:

    print("Prompt:", prompt)

    generated = generate_text(
        prompt,
        max_new_tokens=20,
    )

    print("Generated:", generated)
    print()
```

## 16. Learning Checkpoints

Before moving to a larger text-generation model, understand:

1. **Tokenizer** — converts text to token IDs.
2. **Input / target shift** — input tokens predict the next tokens.
3. **Embedding** — converts token IDs into vectors.
4. **Positional information** — tells the model token positions.
5. **Self-attention** — tokens attend to previous tokens.
6. **Causal masking** — prevents looking at future tokens.
7. **Next-token prediction** — predicts the next vocabulary token.
8. **Autoregressive generation** — feeds each generated token back into the model.
9. **EOS** — tells generation when to stop.

### Translation vs Text Generation

Translation:

`English → Encoder → Decoder → Tamil`

Text generation:

`Previous tokens → Decoder-only Transformer → Next token`

This project is therefore the foundation for understanding **GPT-style decoder-only language models**.
