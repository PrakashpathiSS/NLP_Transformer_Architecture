```python
import re
import numpy as np
import tensorflow as tf
import pandas as pd

from tensorflow.keras.layers import (
    Input,
    Embedding,
    Dense,
    Dropout,
    LayerNormalization,
    MultiHeadAttention,
)
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

print("TensorFlow:", tf.__version__)
```

```python
# ============================================================
# 1. LOAD TEXT SUMMARIZATION DATASET
# ============================================================

dataset = pd.read_csv("transformer_text_summarization_100_rows.csv")

input_texts = dataset["text"].astype(str).tolist()
summary_texts = dataset["summary"].astype(str).tolist()

assert len(input_texts) == len(summary_texts)

print("Training pairs:", len(input_texts))

for i in range(min(5, len(input_texts))):
    print("TEXT:", input_texts[i])
    print("SUMMARY:", summary_texts[i])
    print()
```

```python
# ============================================================
# 2. ADD SOS / EOS TO SUMMARY TARGETS
# ============================================================

summary_sent_SE = [
    f"<sos> {text} <eos>"
    for text in summary_texts
]

print("Input text:", input_texts[0])
print("Target summary:", summary_sent_SE[0])
```

```python
# ============================================================
# 3. TOKENIZERS
# ============================================================

# Both input and output are English in this project.
# The input is the long article/text.
# The output is the short summary.

MAX_INPUT_VOCAB = 10000
MAX_SUMMARY_VOCAB = 5000

input_tokenizer = Tokenizer(
    num_words=MAX_INPUT_VOCAB,
    filters="",
    oov_token="<unk>",
)

summary_tokenizer = Tokenizer(
    num_words=MAX_SUMMARY_VOCAB,
    filters="",
    oov_token="<unk>",
)

input_tokenizer.fit_on_texts(input_texts)
summary_tokenizer.fit_on_texts(summary_sent_SE)

input_sequences = input_tokenizer.texts_to_sequences(input_texts)
summary_sequences = summary_tokenizer.texts_to_sequences(summary_sent_SE)

input_vocab_size = min(
    len(input_tokenizer.word_index) + 1,
    MAX_INPUT_VOCAB,
)

summary_vocab_size = min(
    len(summary_tokenizer.word_index) + 1,
    MAX_SUMMARY_VOCAB,
)

print("Input vocabulary used by model:", input_vocab_size)
print("Summary vocabulary used by model:", summary_vocab_size)
```

```python
# ============================================================
# 4. MAXIMUM SEQUENCE LENGTHS
# ============================================================

max_input_seq_length = max(
    len(x) for x in input_sequences
)

max_output_seq_length = max(
    len(x) for x in summary_sequences
)

print("Max input text length:", max_input_seq_length)
print("Max summary length:", max_output_seq_length)
```

```python
# ============================================================
# 5. PADDING
# ============================================================

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_input_seq_length,
    padding="post",
    truncating="post",
)

output_sequences = pad_sequences(
    summary_sequences,
    maxlen=max_output_seq_length,
    padding="post",
    truncating="post",
)

print("Input shape:", input_sequences.shape)
print("Output shape:", output_sequences.shape)
```

```python
# ============================================================
# 6. DECODER INPUT / TARGET
# ============================================================

# Teacher forcing:
#
# Decoder input:
#   <sos> AI is helping healthcare ...
#
# Decoder target:
#   AI is helping healthcare ... <eos>

decoder_input_sequences = output_sequences[:, :-1]
decoder_target_sequences = output_sequences[:, 1:]

print("Encoder input:", input_sequences.shape)
print("Decoder input:", decoder_input_sequences.shape)
print("Decoder target:", decoder_target_sequences.shape)

# Ignore PAD tokens when calculating the loss/accuracy.
sample_weight = (decoder_target_sequences != 0).astype("float32")
```

```python
# ============================================================
# 7. POSITIONAL EMBEDDING
# ============================================================

class PositionalEmbedding(tf.keras.layers.Layer):
    def __init__(self, vocab_size, sequence_length, d_model, **kwargs):
        super().__init__(**kwargs)
        self.token_embedding = Embedding(vocab_size, d_model, mask_zero=True)
        self.position_embedding = Embedding(sequence_length, d_model)
        self.sequence_length = sequence_length

    def call(self, inputs):
        length = tf.shape(inputs)[-1]
        positions = tf.range(start=0, limit=length, delta=1)
        positions = self.position_embedding(positions)
        tokens = self.token_embedding(inputs)
        return tokens + positions
```

```python
# ============================================================
# 8. TRANSFORMER ENCODER BLOCK
# ============================================================

# Each encoder block contains:
# 1. Multi-Head Self-Attention
# 2. Residual connection + LayerNorm
# 3. Feed-Forward Network
# 4. Residual connection + LayerNorm

class TransformerEncoderBlock(tf.keras.layers.Layer):
    def __init__(self, d_model, num_heads, ff_dim, dropout=0.1, **kwargs):
        super().__init__(**kwargs)
        self.attention = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads,
            dropout=dropout,
        )
        self.ffn = tf.keras.Sequential([
            Dense(ff_dim, activation="relu"),
            Dense(d_model),
        ])
        self.norm1 = LayerNormalization(epsilon=1e-6)
        self.norm2 = LayerNormalization(epsilon=1e-6)
        self.dropout1 = Dropout(dropout)
        self.dropout2 = Dropout(dropout)

    def call(self, x, training=False, attention_mask=None):
        attn_output = self.attention(
            query=x,
            key=x,
            value=x,
            attention_mask=attention_mask,
            training=training,
        )
        x = self.norm1(x + self.dropout1(attn_output, training=training))

        ffn_output = self.ffn(x, training=training)
        x = self.norm2(x + self.dropout2(ffn_output, training=training))
        return x
```

```python
# ============================================================
# 9.TRANSFORMER DECODER BLOCK
# ============================================================

# The decoder has three important operations:
# 1. Causal self-attention — prevents looking at future target tokens.
# 2. Cross-attention — attends to the encoder output.
# 3. Feed-forward network.
    
class TransformerDecoderBlock(tf.keras.layers.Layer):
    def __init__(self, d_model, num_heads, ff_dim, dropout=0.1, **kwargs):
        super().__init__(**kwargs)

        self.self_attention = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads,
            dropout=dropout,
        )

        self.cross_attention = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads,
            dropout=dropout,
        )

        self.ffn = tf.keras.Sequential([
            Dense(ff_dim, activation="relu"),
            Dense(d_model),
        ])

        self.norm1 = LayerNormalization(epsilon=1e-6)
        self.norm2 = LayerNormalization(epsilon=1e-6)
        self.norm3 = LayerNormalization(epsilon=1e-6)

        self.dropout1 = Dropout(dropout)
        self.dropout2 = Dropout(dropout)
        self.dropout3 = Dropout(dropout)

    def call(
        self,
        x,
        encoder_output,
        training=False,
        self_attention_mask=None,
        cross_attention_mask=None,
    ):
        attn1 = self.self_attention(
            query=x,
            key=x,
            value=x,
            attention_mask=self_attention_mask,
            training=training,
        )
        x = self.norm1(x + self.dropout1(attn1, training=training))

        attn2 = self.cross_attention(
            query=x,
            key=encoder_output,
            value=encoder_output,
            attention_mask=cross_attention_mask,
            training=training,
        )
        x = self.norm2(x + self.dropout2(attn2, training=training))

        ffn_output = self.ffn(x, training=training)
        x = self.norm3(x + self.dropout3(ffn_output, training=training))

        return x
```

## 10. Causal mask

At decoder position `t`, the Transformer must not see target tokens after `t`.


```python
# ============================================================
# 10. CAUSAL MASK 
# ============================================================

#At decoder position t, the Transformer must not see target tokens after t

def causal_mask(length):
    # 1 = allowed, 0 = blocked
    i = tf.range(length)[:, None]
    j = tf.range(length)[None, :]
    return tf.cast(i >= j, tf.int32)
```

```python
# ============================================================
# 10. TRANSFORMER
# ============================================================

# Encoder: long-text token IDs → embedding + position → encoder block
# Decoder: summary token IDs → masked self-attention → cross-attention → output logits

class Transformer(tf.keras.Model):
    def __init__(
        self,
        input_vocab_size,
        summary_vocab_size,
        max_input_length,
        max_summary_length,
        d_model=64,
        num_heads=4,
        ff_dim=128,
        dropout=0.1,
        **kwargs,
    ):
        super().__init__(**kwargs)

        self.encoder_embedding = PositionalEmbedding(
            input_vocab_size,
            max_input_length,
            d_model,
        )

        self.decoder_embedding = PositionalEmbedding(
            summary_vocab_size,
            max_summary_length,
            d_model,
        )

        self.encoder = TransformerEncoderBlock(
            d_model, num_heads, ff_dim, dropout
        )

        self.decoder = TransformerDecoderBlock(
            d_model, num_heads, ff_dim, dropout
        )

        self.output_layer = Dense(summary_vocab_size)

    def call(self, inputs, training=False):
        input_tokens, decoder_tokens = inputs

        encoder_x = self.encoder_embedding(input_tokens)
        decoder_x = self.decoder_embedding(decoder_tokens)

        # Padding masks.
        encoder_padding = tf.cast(
            input_tokens != 0, tf.int32
        )[:, None, :]
        decoder_padding = tf.cast(
            decoder_tokens != 0, tf.int32
        )[:, None, :]

        # Encoder self-attention mask: query positions × key positions.
        encoder_mask = tf.cast(
            encoder_padding[:, :, None] * encoder_padding[:, None, :],
            tf.int32,
        )

        # Decoder causal mask.
        target_length = tf.shape(decoder_tokens)[1]
        cmask = causal_mask(target_length)[None, :, :]

        # Decoder padding mask combined with causal mask.
        decoder_mask = cmask * tf.cast(
            decoder_padding[:, None, :],
            tf.int32,
        )

        # Cross-attention: decoder queries may attend to non-padding encoder keys.
        cross_mask = tf.cast(
            encoder_padding[:, None, :],
            tf.int32,
        )

        encoder_output = self.encoder(
            encoder_x,
            training=training,
            attention_mask=encoder_mask,
        )

        decoder_output = self.decoder(
            decoder_x,
            encoder_output,
            training=training,
            self_attention_mask=decoder_mask,
            cross_attention_mask=cross_mask,
        )

        return self.output_layer(decoder_output)
```

```python
# ============================================================
# 11. BUILD THE SMALL SUMMARIZATION TRANSFORMER
# ============================================================

D_MODEL = 64
NUM_HEADS = 4
FF_DIM = 128
DROPOUT = 0.1

transformer = Transformer(
    input_vocab_size=input_vocab_size,
    summary_vocab_size=summary_vocab_size,
    max_input_length=max_input_seq_length,
    max_summary_length=max_output_seq_length - 1,
    d_model=D_MODEL,
    num_heads=NUM_HEADS,
    ff_dim=FF_DIM,
    dropout=DROPOUT,
)

# Build once.
dummy_input = tf.constant(input_sequences[:1], dtype=tf.int32)
dummy_decoder = tf.constant(decoder_input_sequences[:1], dtype=tf.int32)
dummy_logits = transformer([dummy_input, dummy_decoder])

transformer.summary()
print("Logits shape:", dummy_logits.shape)
```

```python
# ============================================================
# 12. COMPILE
# ============================================================

# Targets are integer token IDs, so use sparse categorical crossentropy.
# The model returns logits, not softmax probabilities.

loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=True
)

transformer.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss=loss_fn,
    weighted_metrics=["accuracy"],
)
```

```python
# ============================================================
# 13. TRAIN
# ============================================================

EPOCHS = 50
BATCH_SIZE = 5

history = transformer.fit(
    [input_sequences, decoder_input_sequences],
    decoder_target_sequences,
    sample_weight=sample_weight,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    shuffle=True,
    verbose=1,
)
```

```python
# ============================================================
# 14. INFERENCE: TEXT SUMMARIZATION
# ============================================================

# During inference, the decoder starts with <sos>
# and generates the summary one token at a time until <eos>.

sos_id = summary_tokenizer.word_index["<sos>"]
eos_id = summary_tokenizer.word_index["<eos>"]

def summarize(text, max_length=None):
    input_ids = input_tokenizer.texts_to_sequences([text])

    input_ids = pad_sequences(
        input_ids,
        maxlen=max_input_seq_length,
        padding="post",
        truncating="post",
    )

    decoder_ids = [sos_id]

    if max_length is None:
        max_length = max_output_seq_length - 1

    for _ in range(max_length):
        decoder_input = pad_sequences(
            [decoder_ids],
            maxlen=max_output_seq_length - 1,
            padding="post",
            truncating="post",
        )

        logits = transformer(
            [
                tf.constant(input_ids, dtype=tf.int32),
                tf.constant(decoder_input, dtype=tf.int32),
            ],
            training=False,
        )

        step = len(decoder_ids) - 1
        next_id = int(tf.argmax(logits[0, step]).numpy())

        if next_id == eos_id or next_id == 0:
            break

        decoder_ids.append(next_id)

    index_to_word = {
        idx: word
        for word, idx in summary_tokenizer.word_index.items()
        if idx < summary_vocab_size
    }

    words = []

    for idx in decoder_ids[1:]:
        if idx == eos_id:
            break
        words.append(index_to_word.get(idx, "<unk>"))

    return " ".join(words)
```

```python
# ============================================================
# 15. TEST KNOWN TRAINING EXAMPLES
# ============================================================

for i in range(min(5, len(input_texts))):
    print("TEXT:")
    print(input_texts[i])
    print()
    print("EXPECTED SUMMARY:")
    print(summary_texts[i])
    print()
    print("PREDICTED SUMMARY:")
    print(summarize(input_texts[i]))
    print("\n" + "=" * 80 + "\n")
```

```python
# ============================================================
# 16. TEST COMPLETELY NEW TEXT
# ============================================================

new_texts = [
    "Researchers have created a new battery that can charge electric cars much faster than current batteries. The technology may reduce charging time and make electric vehicles more convenient for daily travel. Scientists are continuing to test its safety, lifetime, and performance before commercial production.",

    "A company has introduced an AI assistant that can analyze documents, answer questions, and create short summaries. The system uses a Transformer model to understand the relationship between words and generate useful responses. The company plans to add support for more languages in the future.",
]

for text in new_texts:
    print("NEW TEXT:")
    print(text)
    print()
    print("GENERATED SUMMARY:")
    print(summarize(text))
    print("\n" + "=" * 80 + "\n")
```

## 17. Important learning checkpoints

Before moving to a larger summarization dataset, understand these points:

1. **Input** = long text/article.
2. **Target** = short summary.
3. The encoder reads the long input.
4. The decoder receives `<sos>` and generates the summary token by token.
5. **Self-attention** helps the encoder understand relationships inside the article.
6. **Cross-attention** lets the decoder look at the encoded article while generating the summary.
7. During training, teacher forcing is used.
8. During inference, the predicted token is fed back to generate the next token.
9. `sparse_categorical_crossentropy` keeps the target data memory-efficient.
10. Training accuracy can become very high on this tiny dataset because the model can memorize the training examples. A genuinely new article is the important test.


```python
## Translation → Summarization: what changed?

The Transformer architecture is essentially the same.

| Translation project | Summarization project |
|---|---|
| English input | Long English text |
| Tamil target | Short English summary |
| English tokenizer | Input-text tokenizer |
| Tamil tokenizer | Summary tokenizer |
| `translate()` | `summarize()` |
| English → Tamil | Long text → Short text |

So we are **not changing the Transformer architecture**. We are changing the **task and dataset**.
```
